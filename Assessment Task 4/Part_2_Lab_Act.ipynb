```python
def reasoning_system():
    exceptions = ["penguin", "ostrich"]
    bird_list = ["penguin", "ostrich", "eagle", "dove", "pigeon"]
    animal = input("Input: ").lower()
    if animal in bird_list:
        print(f"Reasoning: {animal.capitalize()}s are birds.")
        if animal in exceptions:
            print(f"However, {animal}s do not fly.")
            print(f"Conclusion: {animal.capitalize()}s cannot fly.")
        else:
            print(f"Conclusion: {animal.capitalize()}s can fly.")
    else:
        print(f"Reasoning: {animal.capitalize()}s are not birds.")
        print(f"Conclusion: {animal.capitalize()}s cannot fly.")

reasoning_system()
```
